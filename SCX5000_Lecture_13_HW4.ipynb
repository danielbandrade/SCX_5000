```python
import math
import matplotlib.pyplot as plt
import random
```

```python
voters = []

N = 200

total_time = 500

opinions = []

for i in range(N):
  voters.append(random.choice([-1, 1]))

opnion = 0
for i in range(N):
    opnion += voters[i]
opinions.append(opnion/N)

for t in range(1,total_time):
  to_copy = random.choice(range(N))
  to_be_copied = random.choice(range(N))
  voters[to_be_copied] = voters[to_copy]
  opnion = 0
  for i in range(N):
    opnion += voters[i]
  opinions.append(opnion/N)
```

```python
plt.plot(range(total_time), opinions)
plt.grid(True, linestyle=':', alpha=0.6)
plt.ylim(-1, 1) # Set limits to clearly show the range [-1, 1]
plt.xlabel("Time")
plt.ylabel("Opinion")
plt.title("Opinion over Time")
plt.show()
```
